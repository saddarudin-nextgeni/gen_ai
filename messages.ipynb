# Messages

#### Messages are the fundamental unit of context for models in LangChain. They represent the input and output of models, carrying both the content and metadata needed to represent the state of a conversation when interacting with an LLM. Messages are objects that contain:
1. Role that identifies the message type (e.g, system, user)
2. Content that represent the actual content of the message (like text, images, audio, documents etc)
3. Metadata for optional fields such as response information, message IDs, and token usage.

#### LangChain provides a standard message type that works across all model providers, ensuring consistent behavior regardless of the model being called.

```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ['COHERE_API_KEY'] = os.getenv('COHERE_API_KEY')
```

```python
from langchain.chat_models import init_chat_model

llm = init_chat_model('command-a-03-2025')

llm.invoke("Hello, how are you?")
```

## Text Prompts

#### Text prompts are strings - ideal for straight forward generation tasks where you don't need to retain conversation history
#### Use when: 
1. You have a single standalone request
2. You don't need conversation history
3. You want minimal code complexity

```python
response = llm.invoke('What is LangChain?')
print(response.content)
```

## Message Prompts

#### Alternatively you can pass in a list of messages to the model by providing a list of message objects.
#### Message Types:
1. System Message - Tells the model how to behave and provide context for interactions.
2. Human Message - Represents user input and interactions with the model.
3. AI Message - Responses generated by the model, including text content, tool calls, and metadata.
4. Tool Message - Represents the output of tool calls

```python
from langchain.messages import AIMessage, SystemMessage, HumanMessage, ToolMessage

messages = [
    SystemMessage("You are an expert in poetry and you lived your previous life as a poet."),
    HumanMessage("Write a 4 line poetry on 9 to 5 job.")
]

response = llm.invoke(messages)
response
```

```python
print(response.content)
```

```python
messages = [
    SystemMessage("You are a helpful assistant."),
    HumanMessage('Can you help me?'),
    AIMessage("I'd be happy to help you with that question."),
    HumanMessage("What is 2+2?")
]

response = llm.invoke(messages)
print(response.content)
```

```python
# If a model makes tool call this is how it generates AIMessage
AIMessage = AIMessage(
    content=[],
    tool_calls = [{
        'name':'get_weather',
        'args':{'location':'Dadu'},
        'id':'call_123'
    }]
)

weather_result = "Sunny, 72'F"
ToolMessage = ToolMessage(
    content=weather_result,
    tool_call_id = 'call_123'
)

messages = [
    HumanMessage("What's the weather in Dadu"),
    AIMessage,
    ToolMessage
]

response = llm.invoke(messages)
response
```
