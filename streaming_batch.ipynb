## Streaming & Batch

### Most models can stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience, particularly for longer responses. Calling stream() returns an iterator that yields output chunks as they are produced. We can use loop to process each chunk in real time.

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ['GOOGLE_API_KEY'] = os.getenv('GOOGLE_API_KEY')
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model='gemini-2.5-flash-lite')
response = llm.invoke('Hi!')
print(response.content)
```

```python
for chunk in llm.stream('Write 10 line poem on Machine Learning.'):
    print(chunk.text,flush=True)
```

### Using Batch to process multiple inputs parallely

```python
responses = llm.batch([
    "What is weather like in Karachi?",
    "How do airplanes fly?",
    "What is gen ai?"
],
config={
    'max_concurrency':5 #limit to 5 parallel calls
}
)

responses
```
